# Operators & Conditional Statements

Performance Category Program:

6a 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

```python
score_input = input("Enter your score (0 to 10): ")

try:
    score = float(score_input)

    if score < 0 or score > 10:
        print("Invalid input: Please enter a score between 0 and 10 (inclusive).")
    elif score > 7:
        print("Above Average: Great job! Keep up the strong performance.")
    elif score >= 4:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average: Need to Improve your performance, consistent practice will lead to better results.")

except ValueError:
    print("Invalid input: Please enter a numeric value between 0 and 10.")
```
